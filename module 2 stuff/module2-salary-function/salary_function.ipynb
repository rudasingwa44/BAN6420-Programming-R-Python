# Module 2 Assignment: Salary Function
Imports the SF salary data, looks up employees by name, and exports their details to a zipped "Employee Profile" folder.

## Step 1: Import the salary data

```python
import os
import zipfile
import pandas as pd

#step 1 : I'm importing the salary data
try:
    df = pd.read_csv('Total.csv', low_memory=False)
    print(df.shape)
except FileNotFoundError:
    print ("Error: Total.csv not found in this folder.")

#I'm making all names same format so searches ignore capital latters and extra spaces
df['EmployeeName'] = df['EmployeeName'].str.strip().str.upper()
df.head()
```

## Step 2: Employee function (returns details as a list of dictionaries)

```python
#step2 is a function that returns an employee's details as a list of dictionaries 
def get_employee (employee_name):
    if not isinstance (employee_name,str) or employee_name.strip() == "":
        raise ValueError ("Please enter the employee name as text.")
    
    employee_name = employee_name.strip().upper()
    filtered_rows = df[df["EmployeeName"] == employee_name]

    if filtered_rows.empty:
        raise ValueError (employee_name + "was not found.")

    return filtered_rows.to_dict("records")
```

## Step 3: Error handling tests (valid, unknown, empty and non-text names)

```python
# Error handling: test good names, an unknown name, an empty name and a number
for test_name in ["Albert Pardini", "  gary jimenez ", "John Nobody", "", 123]:
    try:
        records = get_employee(test_name)
        print (test_name , "->" , len(records) , "record(s)")
    except ValueError as error:
        print("Error:", error)
```

## Step 4: Export to CSV and zip into "Employee Profile"

```python
# Step 3: Export one employee to CSV inside a zipped "Employee Profile" folder

try:
    employee_data = get_employee("Albert Pardini")

    os.makedirs ("Employee Profile", exist_ok=True)
    csv_path =  "Employee Profile/employee_profile.csv"
    pd.DataFrame (employee_data).to_csv(csv_path,index=False)

    with zipfile.ZipFile("Employee Profile.zip","w") as zipf:
        zipf.write(csv_path)

    print ("saved and zipped" , employee_data[0]["EmployeeName"], "profile!")

except ValueError as error:
    print("Export failed:" , error)
```
