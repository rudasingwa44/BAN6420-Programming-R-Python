```python
class Agent:
    def __init__(self,name,aht):
        self.name = name
        self.aht = aht

    def describe(self):
        return f"{self.name} has aht of {self.aht}s"

ana = Agent ("Ana",300)
print (ana.describe())
print(ana.name)
```

```python
class TeamLead (Agent):
    def __init__(self,name,aht,team):
        super().__init__(name,aht)
        self.team = team
```

```python
class BankAccount:
    def __init__(self,owner,balance):
        self.owner = owner
        self.__balance = balance

    def deposit(self,amount):
        if amount > 0:
            self.__balance += amount

    def get_balance(self):
        return self.__balance
```

```python
class Call:
    def handle(self): return "Answer the phone"
class Email:
    def handle(self): return "Reply in writing"

for contact in [Call(),Email()]:
    print(contact.handle())
```
