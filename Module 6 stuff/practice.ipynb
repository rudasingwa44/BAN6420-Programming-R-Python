```python
from tensorflow import keras
from tensorflow.keras import layers

# Load: 60,000 training + 10,000 test images
(x_train, y_train), (x_test, y_test) = keras.datasets.mnist.load_data()

# Preprocess: scale pixels 0–255 to 0–1, add the "channel" dimension (1 = greyscale)
x_train = x_train.reshape(-1, 28, 28, 1) / 255.0
x_test = x_test.reshape(-1, 28, 28, 1) / 255.0

# Build the model, layer by layer
model = keras.Sequential([
    layers.Input(shape=(28, 28, 1)),
    layers.Conv2D(32, (3, 3), activation="relu"),   # 32 filters, each 3x3
    layers.MaxPooling2D((2, 2)),                     # halve the size
    layers.Dropout(0.25),                            # switch off 25% during training
    layers.Flatten(),
    layers.Dense(128, activation="relu"),
    layers.Dense(10, activation="softmax"),          # 10 classes → probabilities
])

# Compile: choose how it learns and how it's scored
model.compile(optimizer="adam",
              loss="sparse_categorical_crossentropy",
              metrics=["accuracy"])

# Train
model.fit(x_train, y_train, epochs=5, batch_size=64, validation_split=0.1)
model.evaluate(x_test, y_test)
```
