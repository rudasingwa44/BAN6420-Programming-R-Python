```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("survey_data.csv")
print(df.shape)
df.head()
```

```python
income_by_age = df.groupby("age")["total_income"].mean().sort_values(ascending=False)
income_by_age
```

```python
plt.figure(figsize=(10, 5))
income_by_age.plot(kind="bar", color="steelblue")

plt.title("Average Income by Age (Highest First)")
plt.xlabel("Age")
plt.ylabel("Average Total Income (RWF)")
plt.xticks(rotation=0)
plt.ticklabel_format(style="plain", axis="y")

plt.tight_layout()
plt.savefig("chart1_income_by_age.png", dpi=150)
plt.show()
```

```python
df["gender"].value_counts()
```

```python
expense_cols = ["utilities", "entertainment", "school_fees", "shopping", "healthcare"]

spend_by_gender = df.groupby("gender")[expense_cols].mean()
spend_by_gender
```

```python
chart = spend_by_gender.T

chart.plot(kind="bar", figsize=(10, 5))

plt.title("Average Spending per Category by Gender")
plt.xlabel("Spending Category")
plt.ylabel("Average Amount Spent (RWF)")
plt.xticks(rotation=0)
plt.legend(title="Gender")
plt.ticklabel_format(style="plain", axis="y")

plt.tight_layout()
plt.savefig("chart2_spending_by_gender.png", dpi=150)
plt.show()
```

Females spend the most on school fees (~RWF 138k) and lead in utilities and healthcare. Males and Others spend more on entertainment. Respondents identifying as Other report no school fees but high shopping, suggesting a younger group without dependants.
