```python
import matplotlib.pyplot as plt

hours = [8, 9, 10, 11, 12]
calls = [120, 340, 410, 380, 290]

plt.bar(hours,calls)
plt.xlabel("Hour")
plt.ylabel("Calls")
plt.title("Calls Offerd by hour")
plt.grid(True)
plt.savefig("calls.png")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")                            # built-in sample data

sns.scatterplot(data=tips, x="total_bill", y="tip")
plt.show()
```
